Question 1 : What is Simple Linear Regression (SLR)? Explain its purpose.

 Answer : Simple Linear Regression (SLR) is a supervised machine learning and statistical technique used to model the linear relationship between a single independent variable (predictor variable, $X$) and a single dependent variable (target variable, $Y$).

Purpose: The primary purpose of SLR is to predict the value of a continuous dependent variable $Y$ based on the value of $X$, and to quantify the strength and direction of the linear association between the two variables.


Question 2: What are the key assumptions of Simple Linear Regression?

 Answer : 1) Linearity: The relationship between the independent variable $X$ and the dependent variable $Y$ is linear.

 2) Independence of Errors: The residuals (errors) are independent of each other with no autocorrelation.

 3) Homoscedasticity: The variance of the residuals remains constant across all values of $X$.

 4) Normality of Errors: The residuals (errors) follow a normal distribution, especially for small sample sizes.


Question 3: Write the mathematical equation for a simple linear regression model and
explain each term.

 Answer : The mathematical equation for a Simple Linear Regression model is:

 $$Y = \beta_0 + \beta_1 X + \epsilon$$

$Y$: Dependent variable (Target variable being predicted).

$X$: Independent variable (Predictor or feature variable).

$\beta_0$: $Y$-intercept (the predicted value of $Y$ when $X = 0$).

$\beta_1$: Slope coefficient (the average change in $Y$ for every one-unit change in $X$).

$\epsilon$: Error term (residual), representing the difference between the actual $Y$ and the value predicted by the model.

Question 5: What is the method of least squares in linear regression?

 Answer : The Method of Least Squares (also known as Ordinary Least Squares or OLS) is an optimization technique used to find the best-fitting straight line for a set of data points.

 It works by minimizing the Sum of Squared Errors (SSE)—the sum of the squared vertical distances between the actual observed values ($Y_i$) and the predicted values ($\hat{Y}_i$):

 $$\text{SSE} = \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2 = \sum_{i=1}^{n} (Y_i - (\beta_0 + \beta_1 X_i))^2$$

Question 7: Name and briefly describe three common evaluation metrics for regression
models.

 Answer : 1) Mean Absolute Error (MAE): The average of the absolute differences between actual and predicted values. It measures average magnitude of errors without considering direction and is robust to outliers.$$\text{MAE} = \frac{1}{n}\sum_{i=1}^{n}\vert{}Y_i - \hat{Y}_i\vert{}$$

 2) Mean Squared Error (MSE): The average of the squared differences between actual and predicted values. By squaring errors, it heavily penalizes larger errors/outliers.$$\text{MSE} = \frac{1}{n}\sum_{i=1}^{n}(Y_i - \hat{Y}_i)^2$$

 3) Root Mean Squared Error (RMSE): The square root of MSE. It brings the error metric back to the same unit as the target variable $Y$, making interpretation easier.$$\text{RMSE} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(Y_i - \hat{Y}_i)^2}$$


 Question 8: What is the purpose of the R-squared metric in regression analysis?

  Answer : The $R^2$ (R-squared / Coefficient of Determination) metric measures the proportion of variance in the dependent variable $Y$ that can be explained by the independent variable(s) $X$ in the regression model.
  
  Purpose: It assesses the goodness-of-fit of a model on a scale from $0$ to $1$ (or $0\%$ to $100\%$). An $R^2$ of $0.85$ means that $85\%$ of the variability in $Y$ is explained by the regression model, while the remaining $15\%$ is due to unexplained variance or noise.

```python
# Answer:
import numpy as np
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([[1], [2], [3], [4], [5]])  # Independent variable
y = np.array([2, 4, 5, 4, 5])            # Dependent variable

# Initialize and fit the model
model = LinearRegression()
model.fit(X, y)

# Print slope and intercept
slope = model.coef_[0]
intercept = model.intercept_

print(f"Slope (b1): {slope:.4f}")
print(f"Intercept (b0): {intercept:.4f}")
```

Question 10: How do you interpret the coefficients in a simple linear regression model?

 Answer : In the simple linear regression equation $\hat{Y} = \beta_0 + \beta_1 X$:

 Slope ($\beta_1$): Represents the expected average change in the dependent variable $Y$ for every 1-unit increase in the independent variable $X$. If $\beta_1 = 3.5$, $Y$ increases by $3.5$ units for every additional unit increase in $X$.Intercept ($\beta_0$): Represents the baseline expected value of the dependent variable $Y$ when the independent variable $X = 0$.
